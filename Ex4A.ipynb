## AIM

To plot a Normal distribution curve using scipy.stats and matplotlib.

## ALGORITHM

**Step 1:** Start the program.

**Step 2:** Import numpy, matplotlib.pyplot and scipy.stats.norm.

**Step 3:** Enable inline plotting using %matplotlib inline.

**Step 4:** Define the mean (mu) and standard deviation (sigma).

**Step 5:** Generate x values using np.linspace from (mu-4σ) to (mu+4σ).

**Step 6:** Compute the probability density function (pdf) using norm.pdf().

**Step 7:** Plot x against pdf as a line plot.

**Step 8:** Fill the area under the curve using plt.fill_between().

**Step 9:** Add title, x-label, y-label, grid and legend; display the plot.

## PROGRAM

_Click the code and press **Shift + Enter** to run it. The output appears just below the code._

```python
%matplotlib inline
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

mu, sigma = 50, 10
x = np.linspace(mu - 4*sigma, mu + 4*sigma, 200)    # 200 x values around the mean
y = norm.pdf(x, mu, sigma)                          # height of the normal curve

plt.figure(figsize=(8, 5))
plt.plot(x, y, color='blue', linewidth=2, label=f'N(mu={mu}, sigma={sigma})')
plt.fill_between(x, y, color='lightgreen', alpha=0.4)  # shade area under the curve
plt.title('Normal Distribution Curve')
plt.xlabel('X'); plt.ylabel('Probability Density')
plt.grid(True); plt.legend(); plt.show()
```
