## AIM

To perform univariate analysis (Frequency, Mean, Median, Mode, Variance, Standard Deviation, Skewness and Kurtosis) on the Pima Indians Diabetes dataset.

## ALGORITHM

**Step 1:** Start the program.

**Step 2:** Import the pandas and scipy.stats packages.

**Step 3:** Load the Pima Indians Diabetes dataset (diabetes.csv) using pd.read_csv().

**Step 4:** Select the columns Age and Glucose into a sub-DataFrame df1.

**Step 5:** Find the frequency of Age values using value_counts().

**Step 6:** Calculate the mean, median and mode of df1 using mean(), median() and mode().

**Step 7:** Calculate the variance and standard deviation using var() and std().

**Step 8:** Calculate the skewness of the columns using skew().

**Step 9:** Calculate the kurtosis of the columns using kurtosis() from scipy.stats.

**Step 10:** Stop the program.



## PROGRAM

_Click the code and press **Shift + Enter** to run it. The output appears just below the code._

```python
import pandas as pd
from scipy.stats import kurtosis

df = pd.read_csv('data/diabetes.csv')
print("First 5 rows:")
print(df.head())

df1 = pd.DataFrame(df, columns=['Age', 'Glucose'])   # keep only two columns

print("Mean:")
print(df1.mean())
print("Median:")
print(df1.median())
print("Mode:")
print(df1.mode().iloc[0])                            # most frequent value (first row)
print("Variance:")
print(df1.var())
print("Standard Deviation:")
print(df1.std())
print("Skewness:")
print(df1.skew())                                    # skewness = how one-sided the data is
print("Kurtosis:")
print("Age     :", kurtosis(df1['Age'], bias=True))  # kurtosis = peakedness of data
print("Glucose :", kurtosis(df1['Glucose'], bias=True))

print("Frequency of Age (top 5):")
print(df1['Age'].value_counts().head(5))             # frequency of each value

print("Maximum Age:", df1['Age'].max())              # Task 1: added line
print("Minimum Age:", df1['Age'].min())              # Task 2: added line
```
