## AIM

To perform Multiple Regression analysis using more than one independent variable on the Pima Indians Diabetes dataset.

## ALGORITHM

**Step 1:** Start the program.

**Step 2:** Import pandas and LinearRegression from sklearn.linear_model.

**Step 3:** Import r2_score from sklearn.metrics.

**Step 4:** Load the Pima Indians Diabetes dataset (diabetes.csv).

**Step 5:** Select Glucose and BloodPressure as independent variables (X) and Age as dependent (y).

**Step 6:** Create a LinearRegression() object.

**Step 7:** Fit the model with X and y using fit().

**Step 8:** Predict the Age for Glucose=150 and BloodPressure=85.

**Step 9:** Display the regression coefficients and the intercept.

**Step 10:** Compute and display the R² score of the model.

**Step 11:** Stop the program.



## PROGRAM

_Click the code and press **Shift + Enter** to run it. The output appears just below the code._

```python
import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score

df = pd.read_csv('data/diabetes.csv')
print("First 5 rows:")
print(df.head())

X = df[['Glucose', 'BloodPressure']].values       # two inputs = multiple regression
y = df['Age']                                     # value to predict

regr = LinearRegression()
regr.fit(X, y)                                    # train the model

predicted_age = regr.predict([[150, 85]])         # predict for a new patient
print("Predicted Age for Glucose=150, BloodPressure=85:", predicted_age)
print("Coefficients:", regr.coef_)                # one coefficient for each input
print("Intercept:", regr.intercept_)
print("R² Score:", r2_score(y, regr.predict(X)))  # R² near 1 = good fit

print(regr.predict([[120, 80]]))                  # Task 1: added line
print(regr.predict([[100, 70]]))                  # Task 2: added line
```
